### 2. `@` и звёздочка. 5 минут

До запуска напишите, существует ли `P * Q`. Под ячейкой: форма `P @ Q` и почему звёздочка на этой паре незаконна.


```python
P = np.arange(6).reshape(2, 3).astype(float)
Q = np.arange(6).reshape(3, 2).astype(float)
print(P @ Q)
print((P @ Q).shape)

H = np.array([[1.0, 2, 3],
              [4, 5, 6]])
print(H * H)

try:
    print(P * Q)
except ValueError as e:
    print(type(e).__name__)
```

### 3. Определитель и обратная. 5 минут

До запуска назовите определитель `M` по правилу креста. Для `S` достаточно сказать «ноль»: строки пропорциональны.

`det(M)` будет очень близко к −2, с хвостом float. Поэтому вторая проверка идёт через `np.allclose`.


```python
M = np.array([[1.0, 2],
              [3, 4]])
print(np.linalg.det(M))
print(np.allclose(M @ np.linalg.inv(M), np.eye(2)))

S = np.array([[1.0, 2],
              [2, 4]])
print(np.linalg.det(S), np.linalg.matrix_rank(S))

try:
    np.linalg.inv(S)
except np.linalg.LinAlgError as e:
    print(type(e).__name__)
```
