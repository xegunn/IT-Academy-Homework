# 13. Выведите количество игр которая сыграла каждая команда в данном сезоне?

```python
#Cезон в бейсболе включает 81 игру дома и 81 игру на выезде, т.е. 162 игры
tg_regular = pd.DataFrame({
    "all_games" : df_regular.home_team.value_counts() + df_regular.away_team.value_counts() 
})
tg_post = pd.DataFrame({
    "all_games" : df_post.home_team.value_counts() + df_post.away_team.value_counts()
})
print(tg_regular,"\n=============\n",tg_post)
```

# 14. Какая команда выиграла наибольшое количество матчей в сезоне? (Будьте внимательны с типом матча в сезоне - season).

```python
#т.к. в наборе имеется только поле "home_team_win" (победы дома), то его на сложить с значениями "home_team_loss", когда комманда играла в гостях
tg_regular["home_wins"] = df_regular.groupby("home_team").home_team_win.sum()
tg_regular["away_wins"] = df_regular.groupby("away_team").home_team_loss.sum()
tg_regular["total_wins"] = tg_regular.home_wins + tg_regular.away_wins

tg_post["home_wins"] = df_post.groupby("home_team").home_team_win.sum()
tg_post["away_wins"] = df_post.groupby("away_team").home_team_loss.sum()
tg_post["total_wins"] = tg_post.home_wins + tg_post.away_wins
print(tg_regular,"\n", tg_post)
```
