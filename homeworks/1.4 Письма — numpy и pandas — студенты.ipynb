# Задачи. NumPy и Pandas на «Письмах»

Короткий срез интернет-магазина, не все 12 000 клиентов. Приёмы те же, что уже были: массив, маска, пустая почта, одна строка на клиента, верх списка.

Сначала выполните ячейку с данными и больше её не меняйте. Ответ пишите в ячейке под задачей.

## Данные

```python
import numpy as np
import pandas as pd

days = np.array([10, 40, 60, 5, 90, 15, 50, 20])
returned = np.array([1, 0, 0, 1, 0, 1, 0, 1])

clients = pd.DataFrame({
    "client_id": [1, 2, 3, 4, 5, 6, 7, 8],
    "email": ["a@shop.test", "b@shop.test", None, "d@shop.test", "e@shop.test", None, "g@shop.test", "h@shop.test"],
    "days_since": days,
    "returned": returned,
})

orders = pd.DataFrame({
    "client_id": [1, 1, 2, 3, 4, 4, 4, 5, 6, 6, 7, 8, 8],
    "amount": [1000, 800, 400, 800, 700, 600, 200, 400, 500, 700, 300, 900, 100],
})

clients
```

`days` и `returned` стоят в том же порядке, что строки `clients`. Позиция 0 — это `client_id` 1.

`returned` равен 1, если в следующие 30 дней заказ был. `days_since` — сколько дней прошло с последнего заказа.

## NumPy

### 1. Среднее, минимум, максимум

По массиву `days` получите среднее число дней, минимум и максимум. Одной строкой, три числа.

```python
# ответ
print(f"{np.mean(days)}, {np.min(days)}, {np.max(days)}") 
```

### 2. Кто давно не покупал

Сколько людей не покупали 30 дней и дольше. Маска по `days`, потом подсчёт.

```python
# ответдут
len(days[days >= 30])
```

### 3. Доля тех, кто не вернулся

По массиву `returned` посчитайте долю нулей. Это доля людей без заказа в следующие 30 дней. На этом срезе она не обязана совпасть с 18% из лекции.

```python
# ответ
print(f"{len(returned[returned == 0]) * 100  / len(returned)}%")
```

### 4. Три самые давние позиции

Найдите три позиции в `days` с наибольшей давностью, от большей к меньшей. Позиции с нуля, это ещё не `client_id`. Почту здесь не учитывайте.

```python
# ответ
days.sort()
print(f"{days[:-4:-1]}")
```

## Pandas

### 5. Пустая почта

Сколько клиентов без почты и какая это доля от всех строк `clients`.

```python
# ответ
zeroes = clients.email.isna().sum()
all = len(clients)
print(zeroes)
print(f"{zeroes * 100/ all}%")
```

### 6. Кому нажать «отправить»

Оставьте только клиентов с почтой. Среди них возьмите троих с наибольшим `days_since`.

Сверьте с задачей 4. Один человек из той тройки сюда не попадает: письма ему не отправить.

```python
# ответ
cl_mail = clients.dropna(subset = 'email')
cl_mail = cl_mail.sort_values("days_since")[:-4:-1]
cl_mail

```

### 7. Одна строка на клиента

`orders` — журнал, у одного клиента несколько строк. Соберите таблицу: `client_id`, число заказов, сумма `amount`.

```python
# ответ
count_orders = pd.DataFrame({
    "client_id" : orders.client_id.unique(),
    "c_orders" : orders.client_id.value_counts().sort_index(),
    "total_ammount" : orders.groupby("client_id").amount.sum()  
})
count_orders
```

### 8. Новички и возврат

Присоедините к `clients` число заказов из задачи 7. Для каждого числа заказов посчитайте среднюю долю `returned`.

По этому срезу должно быть видно, возвращаются ли люди с одним заказом реже остальных.

```python
# ответ
clients["c_orders"] = count_orders.c_orders.sort_index()
clients
```
